# 1.0 Data Loading and Cleaning
Loads the raw Bank Marketing Data, check quality , clean it, and save processed version.

```python
import pandas as pd
# Raw file URL from GitHub (data/raw/bank-full.csv)
url = "https://raw.githubusercontent.com/gunwes16/data201-project1-classification/refs/heads/main/data/raw/bank-full.csv"

df = pd.read_csv(url, sep=";")   # file is semicolon-separated
print(df.shape)                  # expect (45211, 17)
df.head()
```

```python
df.info()
print("Blank values:", df.isna().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print((df == "unknown").sum()[lambda s: s > 0])
```

```python
print(df["y"].value_counts())
print(df["y"].value_counts(normalize=True).round(3))
```

```python
clean = df.copy()

clean = clean.drop(columns="duration")

clean["prev_contacted"] = (clean["pdays"] != -1).astype(int)

clean["y"] = (clean["y"] == "yes").astype(int)

print(clean.shape)
```

```python
clean.to_csv("bank-clean.csv", index=False)

from google.colab import files
files.download("bank-clean.csv")
```
